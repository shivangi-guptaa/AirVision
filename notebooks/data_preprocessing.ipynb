```python
import pandas as pd

df = pd.read_csv("../delhi-aqi-prediction-using-ml/data/final_aqi_dataset.csv")
print(df.columns)
df_delhi = df[df["city"] == "Delhi"]
print(df_delhi["city"].unique())
```

```python
df_delhi.to_csv("delhi_only.csv", index=False)
```

```python
df["city"] = df["city"].str.strip()
```

```python
df_delhi.groupby("datetime").size().sort_values(ascending=False).head()
```

```python
df_delhi = df_delhi.groupby("datetime").agg({
    "pm25": "mean",
    "pm10": "mean",
    "no2": "mean",
    "so2": "mean",
    "co": "mean",
    "o3": "mean",
    "aqi": "mean",
    "hour": "first",
    "day": "first",
    "month": "first",
    "dayofweek": "first"
}).reset_index()
```

```python
print(df_delhi.shape)
```

```python
print(df_delhi["datetime"].min())
print(df_delhi["datetime"].max())
```

```python
date_range = pd.date_range(
    start=df_delhi["datetime"].min(),
    end=df_delhi["datetime"].max(),
    freq="H"
)

print("Expected:", len(date_range))
print("Actual:", len(df_delhi))
```

```python
df_delhi["datetime"] = pd.to_datetime(df_delhi["datetime"])
df_delhi = df_delhi.sort_values("datetime")
print(df_delhi["datetime"].duplicated().sum())
df_delhi = df_delhi.set_index("datetime").asfreq("H")
print(df_delhi.shape)
df_delhi = df_delhi.interpolate()
df_delhi = df_delhi.reset_index()
```

```python
df_delhi.to_csv("delhi_processed.csv", index=False)
```
