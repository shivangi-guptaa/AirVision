```python
import pandas as pd

df = pd.read_csv("../delhi-aqi-prediction-using-ml/data/final_aqi_dataset.csv")
print(df.columns)
df_delhi = df[df["city"] == "Delhi"]
print(df_delhi["city"].unique())
```

```python
df_delhi.groupby("datetime").size().sort_values(ascending=False).head()
```

```python
df_delhi = df_delhi.groupby("datetime").agg({
    "pm25": "mean",
    "pm10": "mean",
    "no2": "mean",
    "so2": "mean",
    "co": "mean",
    "o3": "mean",
    "aqi": "mean",
    "hour": "first",
    "day": "first",
    "month": "first",
    "dayofweek": "first"
}).reset_index()
```

```python
df_delhi = df_delhi.sort_values("datetime")
```

```python
df_delhi.shape[0]
```

```python
df_delhi["datetime"] = pd.to_datetime(df_delhi["datetime"])
(df_delhi["datetime"].max() - df_delhi["datetime"].min()).days
```

```python
total_hours = (df_delhi["datetime"].max() - df_delhi["datetime"].min()).total_seconds() / 3600
print(total_hours)
```

```python
full_range = pd.date_range(start=df_delhi["datetime"].min(),
                           end=df_delhi["datetime"].max(),
                           freq='H')

missing = full_range.difference(df_delhi["datetime"])

print(len(missing))
```
